## **Part 3: Types and tokens**

To practice the theoretical contents, this part will analyze the types and tokens of the loaded article. The Type-Token Ratio is interpreted as a value between 0 and 1, where higher values indicate greater lexical diversity.

In this exercise you will also encounter the `Counter` library that is mainly used to count the frequency of elements in an iterable, such as lists, strings, and tuples.


👋 ⚒ Exercise 6: Complete the next code cell to calculate the number of tokens and types as well as the Type-Token Ratio (TTR), which is number of types / number of tokens. At which unique token frequency do the words get semantically interesting? How does the TTR value change when you cut the length of the article in half?

```
import re
from collections import Counter

# Lowercasing and keeping only letters and apostrophes
tokens = re.findall(r"[a-z']+", article.text.lower())

# Counting how often each unique token appears in the text
counter = Counter(tokens)
print(f"Count of unique tokens: {counter}")

n_tokens = None  # TODO: number of tokens
n_types = None   # TODO: number of types
ttr = None       # TODO: type-token ratio

print("tokens:", n_tokens)
print("types: ", n_types)
print("TTR:   ", ttr)
print(counter.most_common(20))
```
